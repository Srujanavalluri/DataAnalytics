### Modifications, Insertion and Deletion in NumPy

```python
# modification - existing element value ni change cheyadam
# Insertion - Array lo new element ni add cheyadam
# Deletion - Array nundi element ni remove cheyadam
```

###### Modification of a NumPy Array

```python
import numpy as np
```

```python
south_sweet = np.array([
    ['Motichoor', 'kalakand', 'jelabi'],
    ['boabbatlu', 'kajji kaya', 'khaja']
])
# Modification
south_sweet[0, 0] = 'payasum'
print(south_sweet)
```

###### Updating a Complete Row in NumPy

```python
# Update the complete row
south_sweet[1] = ['kesari', 'pullareddy', 'aarisalu']
print(south_sweet)
```

###### Updating a Complete Column

```python
# Update the complete column
south_sweet[:, 2] = ['mysurepak', 'bread halwa']
print(south_sweet)
```

###### Inserting a New Row using np.insert()

```python
# Insert operation: New row
south_sweet = np.insert(
    south_sweet,
    1,
    ['Gajarpak', 'Rasmalai', 'Firni'],
    axis=0
)
print(south_sweet)
```

###### Insert Multiple Columns

```python
# Purpose: Used to add new columns at a specific position.
south_sweet = np.insert(south_sweet, 2, 
                        [['kajukatli','ghever','peda'], 
                         ['kajukatli','ghever','peda']], axis=1)

print(south_sweet)
```

###### Delete Columns

```python
# Purpose: Used to remove specific columns.
south_sweet = np.delete(south_sweet, [1,3], axis=1)
print(south_sweet)
```

###### 7. Mathematical & Statistical Operations

```python
# 1. np.sum() → Calculates the total of all values.
# 2. np.min() → Finds the smallest value.
# 3. np.max() → Finds the largest value.
# 4. np.mean() → Calculates the average.
# 5. np.std() → Measures how much values vary from the mean.
# 6. np.var() → Measures the spread of values.
# 7. np.percentile() → Finds the value below a given percentage of data.
# 8. np.quantile() → Similar to percentile, but uses values from 0 to 1.
# 9. np.argmax() → Returns the index of the largest value.
# 10. np.argmin() → Returns the index of the smallest value.
# 11. np.sqrt() → Calculates the square root.
# 12. np.round() → Rounds a number to specified decimal places.
# 13. np.floor() → Rounds a number down.
# 14. np.abs() → Returns the absolute (positive) value.
```

###### Adding a Value to Every Element

```python
# Purpose: Adds 100 to every value.
sweet_price = np.array([[450,600,400],[850,400,250]])
print(sweet_price + 100)
```

###### Sum of Array Elements

```python
np.sum(sweet_price)
```

###### Minimum, Maximum, Mean, Standard Deviation & Variance

```python
print(np.min(sweet_price))
# Purpose: Finds the smallest value.
print(np.max(sweet_price))
# Purpose: Finds the largest value.
print(np.mean(sweet_price))
# Purpose: Calculates the average value.
print(np.std(sweet_price))
# Purpose: Measures how much the values vary from the mean.
print(np.var(sweet_price))
# Purpose: Measures the spread/variation of the values.
```

###### Percentile

```python
# Purpose: Finds the value below which a given percentage of the data falls.
# 50% of the people are below 425
print(np.percentile(sweet_price, 50))
```

###### Quantile

```python
# Purpose: Quantile is similar to percentile, but it uses values from 0 to 1.
# Quantile is a special type of percentile
np.quantile(sweet_price, 0.50)
```

###### Finding the Index of the Maximum Value using np.argmax()

```python
# Purpose: Returns the index of the largest value in the array.
age=[22,16,14,8,20]
np.argmax(sweet_price)
```

###### Finding the Index of the Minimum Value using np.argmin()

```python
# Purpose: Returns the index of the smallest value in the array.
# Returns the index of the smallest value
age=[22,16,500,14,8,20,90,200]
np.argmin(age)
```

###### Square Root using np.sqrt()

```python
# Purpose: Calculates the square root of a number.
np.sqrt(900)
```

###### Rounding a Number using np.round()

```python
# Purpose: Rounds a number to the specified number of decimal places.
pi=3.14
np.round(pi,3)
```

###### Floor Value using np.floor()

```python
# Purpose: Rounds a number down to the nearest integer.
np.floor(pi)
```

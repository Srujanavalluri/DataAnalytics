###### Bar Chart – Spotify Song Popularity

```python
# Bar Chart

song_name = [
    "Baby Shark",
    "Coffee First",
    "Sleep More",
    "Monday Blues",
    "Exam Panic"
]

popularity = [40, 22, 55, 81,94]

plt.bar(song_name, popularity)
```

###### Netflix Movies & Shows – Bar Chart

```python
 # Netflix genres
genre = ["Action", "Comedy", "Drama", "Romance", "Thriller"]

# Number of titles in each genre
count = [25, 30, 28, 18, 22]

# Set figure size
plt.figure(figsize=(6, 4))

# Create bar chart with different colors
plt.bar(genre, count, color=["red", "blue", "green", "pink", "orange"])

# Add values on top of each bar
for i, value in enumerate(count):
    plt.text(i, value + 0.5, str(value), ha="center")

# Add X-axis label
plt.xlabel("Genre")

# Add Y-axis label
plt.ylabel("Count")

# Add chart title
plt.title("Netflix Titles by Genre")

# Display chart
plt.show()
```

###### Horizontal Bar Chart

```python
bakery_items = ["Cake", "Bread", "Cookies", "Donut", "Pastry"]
sales = [120, 80, 60, 90, 70]

plt.barh(bakery_items, sales, color=["pink", "brown", "orange", "yellow", "purple"])

plt.xlabel("Sales")
plt.ylabel("Bakery Items")
plt.title("Bakery Sales")

plt.show()
```

##### Histogram

```python
# USE: histogram works on continues data
# used for show the frequency of data in the particular range
```

```python
import numpy as np
data = np.random.randn(1000)
plt.hist(data, bins=30, color="skyblue", edgecolor="black")
plt.show()
```

```python
 # Interview Question:
# What is the difference between a Bar Chart and a Histogram?

# Ans:
# 1. Bar chart is used for categorical/discrete data,
#    while histogram is used for continuous numerical data.
#
# 2. Bar chart is mainly used to compare different categories,
#    while histogram is used to understand the distribution/shape of data.
#
# 3. In a bar chart, there is usually a gap between bars,
#    while histogram bars are usually connected.
```

#### 3D charts in Matplotlib

```python
# 1. 3D chart represents data using three axes: X, Y and Z.
# 2. X-axis → Horizontal direction.
# 3. Y-axis → Depth direction.
# 4. Z-axis → Vertical direction.
# 5. 3D charts are useful for showing relationships between 3 variables.
# 6. projection="3d" is used to create a 3D axis.
# 7. Common 3D charts:
#    3D Scatter Plot
#    3D Line Plot
#    3D Bar Chart
#    3D Surface Plot
# 8. 3D charts are mainly used for data visualization and comparison.
# 9. Matplotlib provides 3D plotting through the mplot3d toolkit.
```

```python
#Import Numpy
import numpy as np
# Import Matplotlib
import matplotlib.pyplot as plt
```

```python
# Create 100 evenly spaced values from 0 to 10
z = np.linspace(0, 10, 100) # np.linspace(start, stop, number)
# Display the values
print(z)
```

###### 3D Line Plot Using Sine and Cosine

```python
 # Create 100 evenly spaced values from 0 to 10
z = np.linspace(0, 10, 100)       # Generate 100 evenly spaced values from 0 to 10
# Calculate sine values for X-axis
x = np.sin(z)                     # Generate sine values in the range -1 to +1
# Calculate cosine values for Y-axis
y = np.cos(z)                     # Generate cosine values in the range -1 to +1
# Create a figure
fig = plt.figure()                 # Creates the main canvas/window for the chart
# Create a 3D axis
ax = fig.add_subplot(111, projection="3d")  # Creates a 3D plotting area inside the figure
# Create a 3D line
ax.plot(x, y, z)                   # Plots the line using X, Y and Z values
```

###### # 3D Scatter Plot Using Sine and Cosine

```python
# 3D Scatter Plot Using Sine and Cosine
# Create 100 evenly spaced values from 0 to 10
z = np.linspace(0, 10, 100)       # Generate 100 evenly spaced values from 0 to 10
# Calculate sine values for X-axis
x = np.sin(z)                     # Generate sine values in the range -1 to +1
# Calculate cosine values for Y-axis
y = np.cos(z)                     # Generate cosine values in the range -1 to +1
# Create a figure
fig = plt.figure()                # Creates the main canvas/window for the chart
# Create a 3D axis
ax = fig.add_subplot(111, projection="3d")  # Creates a 3D plotting area inside the figure
# Create a 3D scatter plot
ax.scatter(x, y, z)               # Plots individual points using X, Y and Z values
```

###### 3D Scatter Plot

```python
# 3D Scatter Plot Using Random Data
x, y, z = np.random.rand(3, 100)
# Create a figure
fig = plt.figure()                 # Creates the main canvas for the chart
# Create a 3D axis
ax = fig.add_subplot(111, projection="3d")  # Creates a 3D plotting area
# Create a 3D scatter plot using X, Y and Z values
ax.scatter(x, y, z)                # Plots individual points using X, Y and Z values
```

###### 3D Wireframe Plot Using X, Y and Z

```python
# 3D Wireframe Plot Using Meshgrid and Sine Function
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
# Create a 2D coordinate grid
x_new, y_new = np.meshgrid(x, y)
# Calculate Z values
z = np.sin(x_new + 10) * np.sin(y_new + 10)
# Create a figure
fig = plt.figure()
# Create a 3D axis
ax = fig.add_subplot(111, projection="3d")
# Create a 3D wireframe
ax.plot_wireframe(x,y,z)
```

##### 3D Surface Chart Using Plotly

```python
# Plotting 3D Surface Chart Using Plotly
import plotly.graph_objects as go
x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
# Create a 2D coordinate grid
x_new, y_new = np.meshgrid(x, y)
# Calculate Z values
z = np.cos(x_new) + np.cos(y_new)
# Create a 3D surface chart
fig = go.Figure(data=[go.Surface(z=z, x=x_new, y=y_new)])
# Display the chart
fig.show()
```

###### Plotting Convex 3D Surface Chart Using Plotly

```python
# Plotting Convex 3D Surface Chart Using Plotly

import plotly.graph_objects as go

x = np.linspace(-5, 5, 30)
y = np.linspace(-5, 5, 30)
x_new, y_new = np.meshgrid(x, y)

z = x_new**2 + y_new**2

fig = go.Figure(data=[go.Surface(z=z, x=x_new, y=y_new)])
fig.show()
```

###### 3D Scatter Plot Using Random Data in Plotly

```python
# 3D Scatter Plot Using Plotly
x1 = np.random.randint(5, 20, 30)
y1 = np.random.randint(5, 25, 30)
z1 = np.random.randint(5, 30, 30)

x2 = np.random.randint(10, 25, 30)
y2 = np.random.randint(5, 15, 30)
z2 = np.random.randint(15, 20, 30)
fig = go.Figure()
fig.add_trace(go.Scatter3d(
    x=x1, y=y1, z=z1, mode="markers"
))
fig.add_trace(go.Scatter3d(
    x=x2, y=y2, z=z2, mode="markers"
))
fig.show()
```

## 2. Creating NumPy Arrays

###### Creating a NumPy Array of Integers

```python
# Main Point:
# NumPy arrays are called ndarray (N-dimensional array).
```

```python
# Example-1: NumPy array of integers
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
print(type(arr1))
```

###### Creating a NumPy Array of Float Values

```python
# type(arr2) → Tells us the object type → ndarray
# arr2.dtype → Tells us the data type of elements → float64
```

```python
# Example-2: NumPy array of float values
arr2 = np.array([1.2, 2.9, 3.3, 4.9, 5.3])
print(arr2)
print(type(arr2))
print(arr2.dtype)
```

###### NumPy Array with Integer and Float Values

```python
# If integers and floats are present together,
# NumPy converts the integers into floats.
# This is called type promotion.
```

```python
arr3 = np.array([1.2, 2, 3.3, 4.9, 5])
print(arr3)
print(type(arr3))
print(arr3.dtype)
```

###### NumPy Array of Strings

```python
# Example-4: NumPy array of strings
arr3 = np.array(["hello", "Class"])
print(arr3)
print(type(arr3))
print(arr3.dtype)
# U → Unicode string data type
# Unicode → Standard used to represent letters, numbers, symbols, and special characters.
```

###### Creating a 2D NumPy Array

```python
arr4 = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])
print(arr4)
print(type(arr4))
print(arr4.dtype)
print(arr4.ndim)    # ndim is used to check the dimension of an array
```

#### Approach 4: Fill with Mean, Median, or Mode

##### Mode 

```python
# Mode : Most frequent category
# replace the null values with mode if data is categical
# Mode: When to Replace Null Values with Mode
# 1. Mode is the most frequently occurring value.
# 2. It can be used for categorical data, such as City or Education_Level.
# 3. It can also be used for numerical data with repeated values.
# 4. It replaces missing values with the most frequent value.
# 5. Example: If "Hyderabad" occurs most often in the City column,
#  we can use "Hyderabad" to fill missing City values.
```

```python
df.isnull().sum()
```

```python
# Section: Finding the Mode of the Gender Column
# Definition: mode() returns the most frequently occurring value.
# Importance: Helps identify the most common gender in the dataset.

mode_value = df.Gender.mode()
mode_value
```

```python
# Section: Checking the Data Type of Mode
# Definition: type() identifies the data type of a variable.
# Important: mode() returns a Pandas Series, even if there is only one mode.

mode_value = df.Gender.mode()
type(mode_value)
mode_value
```

```python
# Section: Filling Null Values in Gender with Mode
# Definition: fillna() replaces missing Gender values with the most frequent value.
# Importance: Keeps records instead of deleting rows with missing Gender.

df.Gender = df.Gender.fillna(mode_value[0])

# Checking Remaining Null Values
df.isnull().sum()
```

```python
# Section: Finding the Mode of Marriage_Status
# Definition: mode() returns the most frequently occurring value.
# Importance: Helps identify the most common marriage status.

mode_value = df.Marrige_Status.mode(0)
mode_value
```

```python
# Section: Filling Null Values in Marriage Status with Mode
# Definition: fillna() replaces missing values with the most frequent value.
# Importance: Keeps records with missing marriage-status information.

df.Marrige_Status = df.Marrige_Status.fillna(mode_value[0])

# Checking Remaining Null Values
# Definition: Counts the missing values in each column.
# Importance: Verifies whether null values remain after filling.

df.isnull().sum()
```

###### Approach 5: Forward Fill (ffill) and Backward Fill (bfill)

```python
# Approach 5: Handling Null Values Using ffill() and bfill()
# Definition: These methods fill missing values using nearby existing values.
# 1. Forward Fill (ffill):
#    Fills a null value with the previous valid value.
# 2. Backward Fill (bfill):
#    Fills a null value with the next valid value.
# Importance:
# Useful for ordered data, such as dates, time series, and records
# where nearby values are meaningful.

# principle : bfill and ffill try to replace te null values with previous or next valid values.
# bfill : backward fill : propagate the next valid value backward to fill missing values.
# ffill : forward fill : propagate the previous valid value forward to fill missing values.
```

```python
 # fill : NaN , 10 ,10 ,20, 30 ,30 ,30 ,25, 12 ,12 
# bfill : 10, 10, 20, 20, 30, 25, 25, 25, 12, NaN
```

```python
df.isnull().sum()
```

```python
# Section: Filling Missing Order Dates Using Backward Fill (bfill)
# Definition: bfill() fills missing values using the next valid value.
# Importance: Helps fill missing dates when the next record's date is appropriate.
df.Date_of_order = df.Date_of_order.bfill()
# Checking Remaining Null Values
# Definition: Counts missing values in each column.
# Importance: Checks whether any null values remain after backward filling.
df.isnull().sum()
```

######  Appraoch -6 : interpolate method 

```python
# Definition: interpolate() estimates missing numerical values using nearby values.
# Importance: Helps fill missing values when data follows a meaningful pattern.
# Example: Numerical measurements recorded over time.

# Interpolate method estimate the missing value points in a continous series
# interpolate only works with numerical data
# "Sorting" data based on date or anyother importent column (eg., UserID)
# we will make interpolate more meaningfull
```

```python
 # 10, 20, 30, NaN, 50, 60
```

```python
 # Approach 6: Filling Missing Values Using Interpolation
# Definition: interpolate() estimates missing numerical values from nearby values.
# Importance: Helps fill missing values in ordered numerical data.
# Step 1: Sort records by order date
df = df.sort_values('Date_of_order')
# Step 2: Fill missing product order counts using interpolation
df['Product_order_count'] = df['Product_order_count'].interpolate()
# Step 3: Check remaining null values
df.isnull().sum()
```

######  EDA-2: 2. Indexing and Filtering

```python
# EDA-2: 2. Indexing and Filtering
# Definition:
# Indexing selects specific rows or columns.
# Filtering selects rows that satisfy a given condition.
# Importance:
# 1. Helps access required data.
# 2. Helps analyze specific records.
# 3. Helps find data based on conditions.
```

```python
# Displaying Column Names
# Definition: Shows all column names in the dataset.
# Importance: Helps identify columns for indexing and filtering.

df.columns
```

```python
# Selecting Rows and Columns Using loc
# Definition: Selects data using row labels and column names.
# Important: The ending labels are included.

df.loc[1:6, 'age':'education']
```

```python
# Selecting Rows and Columns Using iloc
# Definition: Selects data using integer positions.
# Important: The stop index is excluded.

df.iloc[1:7, 0:4]
```

```python
# Selecting Specific Columns Using iloc
# Definition: Selects rows and specific columns by position.
# Important: Column positions 0 and 3 are selected.

df.iloc[1:7, [0, 3]]
```

```python
# Selecting Specific Columns Using loc
# Definition: Selects rows and columns using labels.
# Important: The ending row label is included.

df.loc[1:6, ['age', 'education']]
```

```python
# Filtering Records Where Age Is Less Than 20
# Definition: Selects rows where the age is below 20.
# Importance: Helps find records that meet a specific condition.

df[df.age < 20]
```

```python
# Filtering Records Where Age Is Between 20 and 40
# Definition: Selects records where age is greater than 20 and less than 40.
# Important: & means AND; both conditions must be true.

df[(df.age > 20) & (df.age < 40)]
```

```python
# Filtering Records Where Age Is Greater Than 20 OR Less Than 40
# Definition: Selects records that satisfy at least one condition.
# Important: | means OR; either condition can be true.

df[(df.age > 20) | (df.age < 40)]
```

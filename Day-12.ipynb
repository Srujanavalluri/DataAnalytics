## 3. Special Types of NumPy Arrays

```python
import numpy as np
```

```python
# Special Types of NumPy Arrays
# NumPy provides special functions to create arrays
# with predefined values such as zeros, ones, a range of numbers,
# and evenly spaced values.
# Important functions:
# 1. np.zeros()       → Creates an array filled with 0s
# 2. np.ones()        → Creates an array filled with 1s
# 3. np.full()        → Creates an array filled with a specified value
# 4. np.arange()      → Creates an array with a sequence of numbers
# 5. np.linspace()    → Creates evenly spaced numbers
# 6. np.eye()         → Creates an identity matrix
# 7. np.empty()       → Creates an uninitialized array
```

###### Creating an Array of Zeros using np.zeros()

```python
# Array of zeros
arr_zero = np.zeros(5)
print(arr_zero)
```

###### Creating a 2D Array of Zeros

```python
# Creating a 2D array of zeros
arr_zero = np.zeros((2, 3))
print(arr_zero)
```

###### Creating a 3D Array of Zeros

```python
# Creating a 3D array of zeros
arr_zero = np.zeros((2, 2, 3))
print(arr_zero)
```

###### Creating an Array of Ones using np.ones()

```python
# Array of Ones
arr_ones = np.ones(5)
print(arr_ones)
```

###### Creating a 3D Array of Ones

```python
# Creating a 3D array of ones
arr_ones = np.ones((2, 2, 5))
print(arr_ones)
```

###### Creating a NumPy Array with Specific Values using np.full()

```python
# Initialize the NumPy array with specific values
arr_spe = np.full(5, 100)
print(arr_spe)
```

###### Creating a 2D NumPy Array with a Specific Value using np.full()

```python
arr_spe = np.full((2, 5), 100)
print(arr_spe)
# np.full() → Creates an array filled with a specific value.
# (2, 5) → Creates 2 rows and 5 columns.
# 100 → Fills every element with 100.
```

###### Identity Matrix using np.eye()

```python
# Identity Matrix
# An identity matrix is a square matrix.
# Square matrix → Number of rows = Number of columns.
# Main diagonal values are 1.
# All other values are 0.
identity_matrix = np.eye(4)
print(identity_matrix)
# np.eye(4) → Creates a 4 × 4 identity matrix.
# 4 → Number of rows and columns.
```

### Random package to generate random/synthetic data

```python
# Random Package
# NumPy provides the random module to generate
# random or synthetic data.
# Synthetic Data → Artificially generated data
# that is used for testing, practice, and analysis.
# Important NumPy random functions:
# 1. np.random.rand()      → Random values between 0 and 1
# 2. np.random.random()    → Random values between 0 and 1
# 3. np.random.uniform()   → Random values within a given range
# 4. np.random.normal()    → Random values following a normal distribution
# 5. np.random.choice()    → Randomly selects values from given data
# 6. np.random.seed()      → Makes random results reproducible
```

###### np.random.rand() – Generate Random Values

```python
# np.random.rand() → Generates random values between 0 and 1.
# 5 → Generates 5 random values.
# Values are greater than or equal to 0 and less than 1.
np.random.rand(5)
```

###### np.random.rand(2, 5) – 2D Random Array

```python
# np.random.rand() → Generates random values between 0 and 1.
# (2, 5) → Creates 2 rows and 5 columns.
# Total elements = 2 × 5 = 10.
np.random.rand(2, 5)
```

###### np.random.rand(2, 2, 5, 10) – 4D Random Array

```python
# np.random.rand() → Generates random values between 0 and 1.
# (2, 2, 5, 10) → Creates a 4D array.
# 2 → Number of blocks
# 2 → Rows in each block
# 5 → Columns/groups in each row
# 10 → Elements in each group
np.random.rand(2, 2, 5, 10)
```

###### Uniformly Distributed Random Data

```python
# np.random.uniform() → Generates randomly distributed values
# within a specified range.
# low=1 → Starting value
# high=10 → Upper limit
# size=5 → Generates 5 random values
np.random.uniform(low=1, high=10, size=5)
```

##### 2D Random Array

```python
# np.random.uniform() → Generates uniformly distributed random values.
# low=1 → Lower limit
# high=10 → Upper limit
# size=(2,5) → Creates 2 rows and 5 columns.
# Total elements = 2 × 5 = 10
np.random.uniform(low=1, high=10, size=(2,5))
```

###### Normally Distributed Data

```python
# np.random.normal() → Generates random data
# following a normal distribution.
# loc=5 → Mean of the distribution.
# scale=1.2 → Standard deviation (SD).
# size=(2,5) → Creates 2 rows and 5 columns.
# Total elements = 2 × 5 = 10.

np.random.normal(loc=5, scale=1.2, size=(2,5))
```

### randon() vs normal()

```python
# np.random.random()
# → Generates random values between 0 and 1.
# → Values are uniformly distributed.
# → Mean and standard deviation are not specified.
# → Used for generating general random values.

# np.random.normal()
# → Generates values following a normal distribution.
# → loc → Mean
# → scale → Standard Deviation
# → Values are usually concentrated around the mean.
# → Used for statistical and synthetic data generation.
```

###### Categorical Data

```python
# choice() function is used to randomly select values
# from categorical data.
np.random.seed(42)
# seed(42) → Makes the random output reproducible.
np.random.choice(
    ['paradise', 'spirit', 'pushpa', 'bahubali'],
    size=3
)
# size=3 → Selects 3 random values.
# choice() → Randomly selects values from the given categories.
```

###### Random Choice with Probability

```python
np.random.choice(
    ['Biryani Boss', 'Fauzi Fighter', 'Spirit Star', 'Pushpa Punch', 'Bahubali King'],
    p=[0.2, 0.05, 0.05, 0.1, 0.6],
    size=10
)

# p → Probability of selecting each category.
# size=10 → Generates 10 random values.
# Higher probability → Higher chance of selection.
# All probabilities together should add up to 1.
```

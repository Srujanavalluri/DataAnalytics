###### Line Chart – Monthly Ice Cream Sales

```python
import pandas as pd
import matplotlib.pyplot as plt

dates = pd.date_range(start="2026-1-01", periods=12, freq="ME")

icecream_sales = [120, 150, 130, 180, 200, 220, 250, 230, 190, 170, 140, 160]

plt.figure(figsize=(6, 4))
plt.plot(dates, icecream_sales, marker='o')

for x,y in zip(dates, icecream_sales):
    plt.text(x,y,str(y))

plt.xticks(dates, dates.strftime("%b"), rotation=45)

plt.xlabel("Month")
plt.ylabel("Ice Cream Sales")
plt.title("Monthly Ice Cream Sales")
plt.grid()
plt.show()
```

### Muliple lines chart

```python
# USE: compare two or more trends side by side 
```

```python
import pandas as pd
import matplotlib.pyplot as plt

dates = pd.date_range(start="2015-01-01", periods=12, freq="YE")

flipkart_sales = [20, 45, 160, 180, 200, 220, 250, 280, 300, 330, 360, 400]
amazon_sales = [150, 170, 190, 10, 30, 50, 280, 10, 340, 70, 0, 3]
myntra_sales = [80, 10, 120, 14, 160, 18, 20, 230, 25, 280, 70, 33]

plt.figure(figsize=(7, 4))

plt.plot(dates, flipkart_sales, marker='o', label="Flipkart")
plt.plot(dates, amazon_sales, marker='*', label="Amazon")
plt.plot(dates, myntra_sales, marker='s', label="Myntra")

plt.xticks(dates, dates.strftime("%Y"), rotation=45)

plt.xlabel("Year")
plt.ylabel("Sales")
plt.title("Yearly Sales Comparison")

plt.legend()
```

### Pie Chart

```python
# Pie Chart : Shows how different categories make up a whole.
# Use         : To show proportions or percentages of a total.
# Importance  : Helps easily understand the share of each category.

# pie char is used on categorical data
# pie chart is used to show proportion of the whole 
# use pie chart when number of  categories are less
```

```python
# moblie market share with pie chart 
import matplotlib.pyplot as plt

phones = ["Samsung", "Apple", "Xiaomi", "OnePlus", "Vivo", "Oppo", "Realme"]
market_share = [25, 20, 15, 10, 10, 10, 10]

plt.pie(market_share, labels=phones, autopct="%1.1f%%")

plt.title("Mobile Phone Market Share")
# plt.savefig("Pie.png")
plt.show()
```

### Dount chart

```python
# USe : Dount is like pie chart with space in the middle 
# This middle space can be further used to difen more information
```

```python
import matplotlib.pyplot as plt

activities = ["Sleep", "Study", "Work", "Exercise", "Others"]
hours = [8, 6, 5, 2, 3]

plt.pie(hours, labels=activities, autopct="%1.1f%%",
        wedgeprops={"width": 0.4})

plt.title("Daily Activities")

plt.show()
```

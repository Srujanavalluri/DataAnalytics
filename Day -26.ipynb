###### EDA 5: Identifying the Duplicate Values

```python
# Definition: duplicated() identifies duplicate rows.
# Importance: Helps find repeated rows in the dataset.
# Point 1: sum() counts the duplicate rows.
# Point 2: 0 means no duplicate rows were found.
# Point 3: A positive number indicates duplicate rows exist.
# Point 4: Does not modify the original DataFrame.

df.duplicated().sum()
```

```python
# Definition: duplicated() identifies repeated rows.
# Importance: Helps find duplicate rows in a dataset.
# Point 1: keep="first" keeps the first occurrence unmarked.
# Point 2: Later occurrences are marked as True.
# Point 3: Displays only the duplicate rows.
# Point 4: Does not modify the original DataFrame.

df[df.duplicated(keep="first")]
```

```python
# Identifies duplicate rows.
# Helps maintain data quality.
# first: Keeps first occurrence.
# last: Keeps last occurrence.
# False: Marks all duplicates.
df[df.duplicated(keep=False)].count()
```

```python
# Creates a copy of the dataset.
# Keeps the original DataFrame unchanged.
# Used to remove duplicates safely.

df2 = df.copy()
```

```python
# Removes duplicate rows from df2.
# Keeps the first occurrence by default.
# inplace=True updates df2 directly.

df2.drop_duplicates(inplace=True)
```

###### EDA 6: Identifying and Handling Outliers

```python
# Definition: Outliers are values that differ greatly from most other data.
# Importance: Outliers can affect data analysis.
# Point 1: Outliers affect mean and variance.
# Point 2: They can increase standard deviation.
# Point 3: They may affect some ML algorithms.
# Point 4: Check outliers before removing them.

# Outlier : the data which is not matching with most of the remaining data
# many of the operation like mean,variance,standard deviation, ml algo cannot outlier
```

```python
# Techniques to Handle Outliers:
# 1. IQR method: Identifies outliers using quartiles.
# 2. Z-score method: Identifies values far from the mean.
# 3. Boxplot: Visualizes potential outliers.
```

```python
# Displays potential outliers in Age
import seaborn as sns
sns.boxplot(data=df, x='Age')
```

```python
# Calculating Q1, Q2, Q3, Q4 and IQR

Q1 = df['Age'].quantile(0.25)
Q2 = df['Age'].quantile(0.50)
Q3 = df['Age'].quantile(0.75)
Q4 = df['Age'].quantile(1.00)

IQR = Q3 - Q1

Lower_bound = Q1 - 1.5 * IQR
Upper_bound = Q3 + 1.5 * IQR
print(Q1,Q2,Q3,Q4,IQR,Lower_bound,Upper_bound)
```

```python
# Printing records of people whose age is greater than the Upper Bound

df[df['Age'] > Upper_bound]
```

```python
# Definition: Identifies outliers using the IQR bounds.
# Importance: Finds unusually low and high age values.
# Point 1: LB is the lower bound.
# Point 2: Upper_bound is the upper bound.

outliers = df[(df['Age'] > Upper_bound) | (df['Age'] < Lower_bound)]
outliers
```

```python
df4 = df[(df['Age'] >= Lower_bound) & (df['Age'] <= Upper_bound)]
df4
```

```python
df.shape
```

```python
# Definition: Finds the middle value of the Age column.
# Importance: The median can replace unusually high ages.

median_age = df['Age'].median()
print(median_age)
```

```python
 # Definition: Replaces outlier ages with the median age.
# Importance: Reduces the effect of extreme age values.
# Point 1: | means OR.
# Point 2: .loc updates selected values.

df.loc[(df['Age'] > Upper_bound) | (df['Age'] < Lower_bound), 'Age'] = median_age
```

### countplot

```python
# A countplot is used to show the number of records in each category.
# COUNT PLOT - IMPORTANT POINTS
# 1. Count plot is used to count the number of records in each category.
# 2. It is mainly used for categorical data.
# 3. X-axis → Categories.
# 4. Y-axis → Number of records (Count).
# 5. sns.countplot() is used to create a count plot.
# 6. It is useful for comparing different categories.
# 7. It is similar to a bar chart, but countplot automatically counts the records.
# 8. We do not need to calculate the count manually.
# 9. It can be used to find which category has the highest or lowest count.
# 10. Commonly used columns: Gender, Day, Category, Smoker, etc.
```

###### Count Plot - Time by Sex

```python
sns.countplot(data=df, x='time', hue = 'sex')
```

### KDE (Kernel Density Estimate plot) /

```python
# KDE - Important Points
# 1. KDE shows the distribution of numerical data.
# 2. It represents data using a smooth curve.
# 3. X-axis → Numerical values.
# 4. Y-axis → Density.
# 5. Higher peak → More data is concentrated in that area.
# 6. It helps understand the shape of the data distribution.
# 7. It can show whether data is symmetric or skewed.
# 8. sns.kdeplot() is used to create a KDE plot.
```

```python
plt.figure(figsize = (3,2))
sns.kdeplot(data = df, x = "total_bill", fill = True)
```

#### Violin Plot

```python
# 1. Violin plot shows the distribution of numerical data.
# 2. It shows the shape of the data distribution.
# 3. Wider area → More data is concentrated there.
# 4. Narrow area → Less data is concentrated there.
# 5. It can show the median and spread of data.
# 6. It is useful for comparing distributions between groups.
# 7. sns.violinplot() is used to create a violin plot.
```

```python
# Violin Plot - Total Bill by Smoker
sns.violinplot(x="smoker", y="total_bill", data=df)
```

### Regression plot

```python
# A regression plot shows the relationship between two numerical variables and adds a best-fit regression line.
# 1. Regression plot shows the relationship between two numerical variables.
# 2. It shows the data points using a scatter plot.
# 3. It adds a regression line (best-fit line).
# 4. The line shows the general trend of the data.
# 5. Upward line → Positive relationship.
# 6. Downward line → Negative relationship.
# 7. sns.regplot() is used to create a regression plot.
# 8. It is useful for understanding correlation and trends.
```

```python
# USE: Used to plot the linear regression line
plt.figure(figsize=(3,2))

sns.regplot(
    x="total_bill",
    y="tip",
    data=df
)
```

### facetgrid

```python
# FacetGrid is used to create multiple plots for different categories in a single figure.
# 1. FacetGrid creates multiple plots in one figure.
# 2. It is useful for comparing different categories.
# 3. col="sex" → Creates separate plots for Male and Female.
# 4. map() → Applies a plot to each category.
# 5. sns.scatterplot → Plot used inside the FacetGrid.
# 6. total_bill → X-axis.
# 7. tip → Y-axis.
# 8. sns.FacetGrid() is used to create a FacetGrid.
```

```python
# USE: To plot multiple charts based on different categories

obj = sns.FacetGrid(df, col="smoker")

obj.map_dataframe(
    sns.scatterplot,
    x="total_bill",
    y="tip"
)

plt.show()
```

### Joint plot

```python
# 1. Joint plot shows the relationship between two variables.
# 2. It combines a main plot with distribution plots.
# 3. The center shows the relationship between X and Y.
# 4. The top shows the distribution of the X variable.
# 5. The side shows the distribution of the Y variable.
# 6. sns.jointplot() is used to create a joint plot.
# 7. It is mainly used for numerical variables.
```

```python
# combine two charts
# Joint Plot - Total Bill vs Tip

sns.jointplot(
    x="total_bill",
    y="tip",
    data=df
)

plt.show()
```

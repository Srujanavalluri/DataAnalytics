#### Seaborn

```python
# SEABORN
# Def: Seaborn is a high-level data visualization library built on top of Matplotlib.
# Seaborn is mainly used for creating statistical charts such as
# Box Plot, KDE Plot, Pair Plot, Heatmap, etc.
# Seaborn provides some built-in datasets for practice and learning.
# Matplotlib provides more detailed control over charts,
# while Seaborn provides simpler and easier ways to create statistical charts.
```

```python
import seaborn as sns
```

```python
# Seaborn Built-in Datasets
print(sns.get_dataset_names())  # Displays the names of datasets available in Seaborn
```

### line chart
#### All the charts of the matplotlib can also be plotted by seaborn

```python
# LINE CHART USING SEABORN
# All the charts that can be created using Matplotlib
# can also be created using Seaborn.
# Seaborn provides simpler functions for creating many
# types of statistical charts.
```

######  Line Chart - Passenger Trends Over Time

```python
# trends over the time 
# On the Chart freatures of seaborn and matploiib can be apply

import matplotlib.pyplot as plt

timeline = df.groupby("year")["passengers"].sum().reset_index()

sns.lineplot(
    x="year",
    y="passengers",
    data=timeline,
    marker="o"
)
plt.title("Passenger Trends Over Time")
plt.grid()
```

```python
# scatter plot

df = sns.load_dataset("tips")
df
```

```python
 # Scatter Plot - Relationship Between Two Variables

sns.scatterplot(x="total_bill", y="tip", data=df)
plt.title("Relationship Between Total Bill and Tip")
```

```python
# Pair Plot - Compare Multiple Variables Together
sns.pairplot(df)
```

```python
# Scatter Plot - Compare Two Variables by Category

plt.figure(figsize=(4,3))

sns.scatterplot(
    x="total_bill",
    y="tip",
    data=df,
    hue="smoker"
)

plt.title("Total Bill vs Tip by Smoker")
```

```python
# Pair Plot - Compare Numerical Variables Using Histograms
plt.figure(figsize=(4,3))
sns.pairplot(df, hue="smoker", kind="hist")
```

```python
# Pair Plot - Compare All Numerical Pairs Using Scatter Plots
sns.pairplot(df, hue="smoker", kind="scatter")
```

#### Heatmap

```python
# HEATMAP - CORRELATION
# Use: Heatmap is used to visualize the correlation between numerical variables.
# Correlation matrix shows the correlation value between each pair of numerical variables.
import pandas as pd
corr = df.corr(numeric_only=True)
print(corr)
```

```python
# Heatmap - Correlation Matrix
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
import pandas as pd
import numpy as np

# Create monthly date range
date_rng = pd.date_range(
    start='2021-01-01',
    end='2026-12-31',
    freq='ME'
)

print(date_rng)
```

```python
# Pick 1000 random dates
random_dates = np.random.choice(
    date_rng,
    size=1000,
    replace=True
)

print(len(random_dates))
```

```python
# Creating DataFrame using random_dates
canteen_df = pd.DataFrame({
    
    'timestamp': pd.to_datetime(random_dates),
    
    'order_id': np.arange(1, 1001),
    # Generate order IDs from 1 to 1000
    
    'food_category': np.random.choice(
        ["Samosa Simba", "Puff Pichha", "Tea Thagubothu",
         "Choki Choki Rowdy", "Bajji Bujji"],
        size=1000
    ),
    
    'price': np.random.randint(10, 500, size=1000),
    
    'quantity': np.random.randint(1, 5, size=1000)
})

canteen_df
```

###### Sorting Time Series Data by Timestamp

```python
# sort_values("timestamp")
# → Sorts the data according to the timestamp.
# reset_index(drop=True)
# → Resets the index after sorting.
# → drop=True removes the old index.
```

```python
# Sort time series data based on timestamp
canteen_df = canteen_df.sort_values("timestamp").reset_index(drop=True)
canteen_df
```

###### Extracting Year, Month and Day from Timestamp

```python
# .dt.year  → Extracts the year from the timestamp.
# .dt.month → Extracts the month from the timestamp.
# .dt.day   → Extracts the day from the timestamp.
```

```python
# Create new columns with year, month, and day
canteen_df['year'] = canteen_df['timestamp'].dt.year
canteen_df['month'] = canteen_df['timestamp'].dt.month
canteen_df['day'] = canteen_df['timestamp'].dt.day
canteen_df
```

###### Extracting Hour, Minute and Second

```python
# .dt.hour   → Extracts hour
# .dt.minute → Extracts minute
# .dt.second → Extracts second
canteen_df['hour'] = canteen_df['timestamp'].dt.hour
canteen_df['minute'] = canteen_df['timestamp'].dt.minute
canteen_df['second'] = canteen_df['timestamp'].dt.second
canteen_df
```

### Copy in Pandas

```python
# Types of Copy in Pandas
# 1. df1 = df
#    → Gives another name/reference to the same DataFrame.
#    → No new copy is created.
# 2. df2 = df.copy()
#    → Creates a new duplicate copy of the dataset.
#    → This is a Deep Copy by default.
# 3. Shallow Copy
#    → Dataset structure is different, but underlying data can be common/shared.
#    → Syntax: df3 = df.copy(deep=False)
```

###### Shallow Copy in Pandas

```python
# pd.DataFrame() → Creates a DataFrame.
# copy(deep=False) → Creates a shallow copy.
# deep=False → The new DataFrame structure is different,
# but the underlying data can be shared with the original.
# original_data → Shallow copy of tourist_data.
```

```python
tourist_data = pd.DataFrame({
    'Place': ['Mumbai', 'Goa', 'Paris', 'Thailand'],
    'package': [10, 50, 200, 100]
})
original_data = tourist_data.copy(deep=False)
tourist_data
```

###### Modifying Data in a Shallow Copy

```python
# Change the package value in the shallow copy
original_data.iloc[0, 1] = 20
tourist_data
```

###### Difference Between Normal Copy and Shallow Copy in Pandas

```python
# Difference between Normal Copy and Shallow Copy

# Normal Copy / Deep Copy:
# df.copy() → Creates an independent copy of the DataFrame.
# Changes in the copied DataFrame normally do not affect the original DataFrame.

# Shallow Copy:
# df.copy(deep=False) → Creates a new DataFrame structure.
# The underlying data can be shared with the original DataFrame.

# Easy Memory:
# Deep Copy    → Independent data
# Shallow Copy → Shared underlying data
```
